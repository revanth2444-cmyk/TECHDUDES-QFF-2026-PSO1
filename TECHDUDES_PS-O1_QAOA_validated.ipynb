**Goal:** Divide the vertices into two groups so that the total weight of edges connecting different groups is maximized.

**Vertices:** 0, 1, 2, 3, 4, 5

**Weighted edges:**

- (0, 4): 4

**Total edge weight:** 27

## AerSimulator Setup

`run_exp()` uses `save_expectation_value` to evaluate $\langle\psi|H|\psi\rangle$ directly on the statevector.  
`run_counts()` appends measurements and returns the shot histogram.

```python
simulator = AerSimulator(method="statevector")


def run_exp(circuit: QuantumCircuit, observable: SparsePauliOp, shots: int = 1024) -> float:
    """Compute expectation value ⟨ψ|H|ψ⟩ using AerSimulator directly."""
    qc = circuit.copy()
    qc.save_expectation_value(operator=observable, qubits=list(range(qc.num_qubits)))
    return float(simulator.run(qc, shots=shots).result().data()["expectation_value"])


def run_counts(circuit: QuantumCircuit, shots: int = 1024) -> dict:
    """Sample the circuit and return measurement counts."""
    qc = circuit.copy()
    for i in range(qc.num_qubits):
        qc.measure(i, i)
    return simulator.run(qc, shots=shots).result().get_counts()


print("AerSimulator ready:", simulator)
```

## Cost Function

The QAOA cost function binds the current angle vector ($\beta$ interleaved with $\gamma$) into the decomposed `QAOAAnsatz` circuit and evaluates the expectation value $\langle\psi(\beta, \gamma)|H|\psi(\beta, \gamma)\rangle$ via `run_exp()`. The optimization loop below caches every evaluated point so its logging callback can reuse the value instead of re-running the circuit.

```python
def cost(
    angles: np.ndarray,
    circuit: QuantumCircuit,
    hamiltonian: SparsePauliOp,
    shots: int,
) -> float:
    """Evaluate ⟨ψ(β, γ)|H|ψ(β, γ)⟩ for the given angle vector."""
    bound_circuit = circuit.assign_parameters(parameters=angles, flat_input=True)
    return run_exp(bound_circuit, hamiltonian, shots=shots)
```

## Problem Definition

MaxCut on a triangle graph: maximize the number of edges cut, equivalent to minimizing $-(x_0\bar{x}_1 + x_1\bar{x}_2 + x_0\bar{x}_2)$.  
In QUBO form this becomes: minimize $-x_0 - x_1 - x_2 + 2x_0x_1 + 2x_1x_2 + 2x_0x_2$.

`build_ising_problem()` converts the QUBO to an Ising Hamiltonian via `QuadraticProgram.to_ising()`.  
`QAOAAnsatz` builds the full $p$-layer QAOA circuit from the Ising operator; `decompose(reps=2)` expands it into native gates.

```python
from typing import Any, Dict, List
from qiskit_optimization import QuadraticProgram


def build_ising_problem(
    variables: List[Dict[str, Any]],
    cost_function: Dict[str, Dict[str, Any]],
    maximization: bool,
):
    var_names = [v["name"] for v in variables if v.get("name")]
    if not var_names:
        raise ValueError("No valid variable names found")

    linear, quadratic = {}, {}

    for _, term in cost_function.items():
        coeff = float(term["coefficient"])
        v1, v2 = term["var1"], term["var2"]

        if v1 == "1" and v2 in var_names:
            linear[v2] = linear.get(v2, 0.0) + coeff
        elif v2 == "1" and v1 in var_names:
            linear[v1] = linear.get(v1, 0.0) + coeff
        elif v1 in var_names and v2 in var_names:
            if v1 == v2:
                linear[v1] = linear.get(v1, 0.0) + coeff
            else:
                quadratic[tuple(sorted((v1, v2)))] = (
                    quadratic.get(tuple(sorted((v1, v2))), 0.0) + coeff
                )
        else:
            raise ValueError(f"Invalid term: {v1}, {v2}")

    qp = QuadraticProgram()
    for name in var_names:
        qp.binary_var(name=name)

    if maximization:
        qp.maximize(linear=linear, quadratic=quadratic)
    else:
        qp.minimize(linear=linear, quadratic=quadratic)

    hamiltonian, offset = qp.to_ising()
    return qp, hamiltonian, var_names


variables = [{'name': 'x₁', 'type': 'Binary'}, {'name': 'x₂', 'type': 'Binary'}, {'name': 'x₃', 'type': 'Binary'}, {'name': 'x₄', 'type': 'Binary'}, {'name': 'x₅', 'type': 'Binary'}, {'name': 'x₆', 'type': 'Binary'}]
cost_function = {'t1': {'coefficient': -9, 'var1': 'x₁', 'var2': '1'}, 't2': {'coefficient': -6, 'var1': 'x₂', 'var2': '1'}, 't3': {'coefficient': -7, 'var1': 'x₃', 'var2': '1'}, 't4': {'coefficient': -15, 'var1': 'x₄', 'var2': '1'}, 't5': {'coefficient': -12, 'var1': 'x₅', 'var2': '1'}, 't6': {'coefficient': -5, 'var1': 'x₆', 'var2': '1'}, 't7': {'coefficient': 2, 'var1': 'x₁', 'var2': 'x₃'}, 't8': {'coefficient': 8, 'var1': 'x₁', 'var2': 'x₄'}, 't9': {'coefficient': 8, 'var1': 'x₁', 'var2': 'x₅'}, 't10': {'coefficient': 6, 'var1': 'x₂', 'var2': 'x₅'}, 't11': {'coefficient': 6, 'var1': 'x₂', 'var2': 'x₆'}, 't12': {'coefficient': 10, 'var1': 'x₃', 'var2': 'x₄'}, 't13': {'coefficient': 2, 'var1': 'x₃', 'var2': 'x₅'}, 't14': {'coefficient': 8, 'var1': 'x₄', 'var2': 'x₅'}, 't15': {'coefficient': 4, 'var1': 'x₄', 'var2': 'x₆'}}
MAXIMIZATION = False

qp, problem_hamiltonian, var_names = build_ising_problem(
    variables=variables,
    cost_function=cost_function,
    maximization=MAXIMIZATION,
)

ising_hamiltonian, offset = qp.to_ising()
n_qubits = len(var_names)
print(f"Variables : {var_names}")
print(f"Qubits    : {n_qubits}")
print(f"QuadraticProgram:\n{qp}")
print(f"Hamiltonian offset: {offset}")
print(f"Hamiltonian:\n{ising_hamiltonian}")
```

## SPSA Optimizer (optional)

`scipy.optimize`'s derivative-free methods (COBYLA, Nelder-Mead, SLSQP) assume a smooth, unimodal objective, which shot noise violates. SPSA (Simultaneous Perturbation Stochastic Approximation) is the standard choice for noisy/shot-sampled objectives and costs exactly 2 circuit evaluations per iteration regardless of parameter count, so its total budget (`2*maxiter+1`) is known up front. Inlined here from `src/gradients.py` — only used when `OPTIMIZER == "SPSA"`.

```python
class SPSAResult:
    """Minimal scipy.optimize.OptimizeResult-like container."""

    def __init__(self, x, fun, nfev, cost_history):
        self.x = x
        self.fun = fun
        self.nfev = nfev
        self.cost_history = cost_history


def minimize_spsa(
    fun,
    x0,
    maxiter=100,
    a=0.1,
    c=0.1,
    alpha=0.602,
    gamma=0.101,
    seed=None,
):
    """Minimize `fun` with Simultaneous Perturbation Stochastic Approximation.

    Always costs exactly 2 evaluations of `fun` per iteration, regardless of
    len(x0). `alpha`/`gamma` are Spall's standard asymptotically optimal decay
    exponents.
    """
    rng = np.random.default_rng(seed)
    theta = np.asarray(x0, dtype=float).copy()
    cost_history = []
    nfev = 0

    for k in range(maxiter):
        ak = a / (k + 1) ** alpha
        ck = c / (k + 1) ** gamma
        delta = rng.choice([-1.0, 1.0], size=theta.shape)

        f_plus = fun(theta + ck * delta)
        f_minus = fun(theta - ck * delta)
        nfev += 2

        # Average of the two perturbation evals as this iteration's cost --
        # free (no extra evaluation), since theta itself is never evaluated
        # mid-loop.
        cost_history.append((f_plus + f_minus) / 2.0)

        grad = (f_plus - f_minus) / (2.0 * ck) * delta
        theta = theta - ak * grad

    final_fun = fun(theta)
    nfev += 1

    return SPSAResult(x=theta, fun=final_fun, nfev=nfev, cost_history=cost_history)
```

## QAOA Optimization

`scipy.optimize.minimize` drives the classical outer loop for `COBYLA`, `Nelder-Mead`, and `SLSQP`; `minimize_spsa` (above) drives it for `SPSA`. Each evaluated point is cached in `cost_cache` so the logging callback reuses it instead of re-running the circuit — this halves the true circuit-eval count versus re-evaluating on every callback.

Initial angles: $[\beta_1, \ldots, \beta_p, \gamma_1, \ldots, \gamma_p]$.

```python
valid_optimizers = ["COBYLA", "Nelder-Mead", "SLSQP", "SPSA"]
if OPTIMIZER not in valid_optimizers:
    raise ValueError(f"Optimizer {OPTIMIZER} not supported. Choose one of {valid_optimizers}")

MAXITER = 50

# Initial parameter vector: [beta_1, ..., beta_p, gamma_1, ..., gamma_p]
init_angles = [INIT_BETA] * P + [INIT_GAMMA] * P

# Cache every evaluated point so the logging callback below can reuse
# the value instead of re-running the circuit.
cost_history = []
cost_cache = {}


def cost_fn(angles):
    value = cost(angles, circuit, ising_hamiltonian, SHOTS)
    cost_cache[tuple(angles)] = value
    return value


def callback(x):
    key = tuple(x)
    val = cost_cache[key] if key in cost_cache else cost_fn(x)
    cost_history.append(val)
    print(f"  Iteration {len(cost_history):3d} | cost = {val:.6f}")


print("Starting QAOA optimization...")
if OPTIMIZER == "SPSA":
    opt_result = minimize_spsa(cost_fn, init_angles, maxiter=MAXITER or 100)
    cost_history = opt_result.cost_history
else:
    opts = {"maxiter": MAXITER} if MAXITER is not None else None
    opt_result = minimize(
        fun=cost_fn,
        x0=init_angles,
        method=OPTIMIZER,
        callback=callback,
        options=opts,
    )

print(f"\nOptimization finished after {opt_result.nfev} function evaluations.")
print(f"Optimal angles : {opt_result.x}")
print(f"Optimal cost   : {opt_result.fun:.6f}")
```

## `run_qaoa()` — All-in-One Function

Encapsulates the full QAOA pipeline: problem construction → circuit build → optimization → sampling → result dict.

```python
def run_qaoa(
    variables,
    cost_function,
    p: int = 1,
    optimizer_str: str = "COBYLA",
    maximization: bool = False,
    shots: int = 1024,
    init_beta: float = np.pi / 4,
    init_gamma: float = np.pi / 8,
    maxiter: int | None = None,
) -> dict:
    """Run the full QAOA pipeline and return a result dictionary.

    Parameters
    ----------
    variables       : list of dicts with key ``"name"`` for each binary variable
    cost_function   : dict of terms, each with ``"coefficient"``, ``"var1"``, ``"var2"``
    p               : number of QAOA layers (reps)
    optimizer_str   : "COBYLA", "Nelder-Mead", "SLSQP", or "SPSA"
    maximization    : if True, negate the objective for maximization
    shots           : measurement shots per evaluation
    init_beta       : initial β angle for all p layers
    init_gamma      : initial γ angle for all p layers
    maxiter         : max optimizer iterations (SPSA defaults to 100 if None)

    Returns
    -------
    dict with keys:
        ``optimization_result``, ``solution_bitstring``, ``objective_value``,
        ``cost_history``, ``counts``
    """
    valid_optimizers = ["COBYLA", "Nelder-Mead", "SLSQP", "SPSA"]
    if optimizer_str not in valid_optimizers:
        raise ValueError(
            f"Optimizer {optimizer_str} not supported. Choose one of {valid_optimizers}"
        )

    # Build Ising problem
    _qp, _, _var_names = build_ising_problem(
        variables=variables,
        cost_function=cost_function,
        maximization=maximization,
    )
    _hamiltonian, _ = _qp.to_ising()
    _n_qubits = len(_var_names)

    # Build circuit
    _qaoa = QAOAAnsatz(cost_operator=_hamiltonian, reps=p)
    _circuit = QuantumCircuit(_n_qubits, _n_qubits)
    _circuit.compose(_qaoa, inplace=True)
    _circuit = _circuit.decompose(reps=2)

    # Initial angles
    _init_angles = [init_beta] * p + [init_gamma] * p

    # Cache every evaluated point so the logging callback below can reuse
    # the value instead of re-running the circuit.
    _cost_history = []
    _cost_cache = {}

    def _cost_fn(angles):
        value = cost(angles, _circuit, _hamiltonian, shots)
        _cost_cache[tuple(angles)] = value
        return value

    def _cb(x):
        key = tuple(x)
        _cost_history.append(_cost_cache[key] if key in _cost_cache else _cost_fn(x))

    # Optimize
    if optimizer_str == "SPSA":
        _result = minimize_spsa(_cost_fn, _init_angles, maxiter=maxiter or 100)
        _cost_history = _result.cost_history
    else:
        _opts = {"maxiter": maxiter} if maxiter is not None else None
        _result = minimize(
            fun=_cost_fn,
            x0=_init_angles,
            method=optimizer_str,
            callback=_cb,
            options=_opts,
        )

    # Sample final circuit
    _final_circuit = _circuit.assign_parameters(
        parameters=_result.x, flat_input=True
    )
    _counts = run_counts(_final_circuit, shots=shots)

    # Decode solution
    _sol_bitstring = max(_counts, key=_counts.get)
    _bitstring_list = list(map(int, _sol_bitstring[::-1]))
    _obj_value = _qp.objective.evaluate(_bitstring_list)

    return {
        "optimization_result": _result,
        "solution_bitstring": _sol_bitstring,
        "objective_value": _obj_value,
        "cost_history": _cost_history,
        "counts": _counts,
    }
```
